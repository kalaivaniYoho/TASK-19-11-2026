```python

import pandas as pd

# Load cleaned employee data
df = pd.read_csv("clean_data.csv")

# 1. Summary statistics
print("1. SUMMARY STATISTICS")
print(df[["age", "salary"]].agg(
    ["mean", "median", "min", "max", "std"]
).round(2))

# 2. Group comparisons
print("\n2. DEPARTMENT-WISE ANALYSIS")
print(df.groupby("department").agg(
    employees=("employee_id", "count"),
    average_salary=("salary", "mean"),
    average_age=("age", "mean")
).round(2))

print("\nCITY-WISE ANALYSIS")
print(df.groupby("city").agg(
    employees=("employee_id", "count"),
    average_salary=("salary", "mean")
).round(2))

# 3. Top three numeric correlations
print("\n3. TOP 3 CORRELATIONS")
numeric = df[["employee_id", "age", "salary"]]
correlations = numeric.corr()

pairs = correlations.where(
    ~pd.DataFrame(
        [[i == j for j in correlations.columns]
         for i in correlations.index],
        index=correlations.index,
        columns=correlations.columns
    )
).stack()

pairs = pairs[
    [a < b for a, b in pairs.index]
]

top3 = pairs.reindex(
    pairs.abs().sort_values(ascending=False).index
).head(3)

print(top3)

# 4. Five business questions with numeric answers
print("\n4. BUSINESS QUESTIONS")

print("Q1. How many employees are there?")
print(len(df))

print("\nQ2. What is the average salary?")
print(round(df["salary"].mean(), 2))

print("\nQ3. Which department has the most employees?")
print(df["department"].value_counts().head(1))

print("\nQ4. Which city has the highest average salary?")
print(df.groupby("city")["salary"].mean().idxmax())

print("\nQ5. How many salary outliers exist?")
print(df["salary_outlier"].sum())

# 5. Merge department budget data
print("\n5. DEPARTMENT BUDGET MERGE")
budget = pd.read_csv("department_budget.csv")

merged = pd.merge(
    df,
    budget,
    on="department",
    how="left"
)

print(merged[[
    "employee_id", "department", "salary", "budget"
]].to_string(index=False))

print("\nTASK 2 COMPLETED")
```
